```python
import numpy as np

# Define matrix A and vector b
A = np.array([[2.0, 1.0],
              [1.0, 3.0]])

b = np.array([1.0, 2.0])

# Display Matrix A
print("Matrix A:")
print(A)

# Transpose of A
print("\nTranspose A^T:")
print(A.T)

# Inverse of A
print("\nInverse A^-1:")
print(np.linalg.inv(A))

# Eigenvalues and eigenvectors
eigvals, eigvecs = np.linalg.eig(A)

print("\nEigenvalues:")
print(np.round(eigvals, 4))

print("\nEigenvectors:")
print(np.round(eigvecs, 4))

# Solve Ax = b
x = np.linalg.solve(A, b)

print("\nSolve A x = b:")
print(np.round(x, 4))
```
